## (ii) Wigner $3$-$j$ symbol calculator
### To evaluate: $\left( \begin{array}{ccc}
j_{1} & j_{2} & j\\
m_{1} & m_{2} & m \end{array} \right)$
### Call the method:
`am.symb3j(j1, j2, j, m1, m2, m)`

The Wigner $3j$-symbol may be evaluated analytically using the following relationship:

\begin{equation}
\left( \begin{array}{ccc}
j_{1} & j_{2} & j\\
m_{1} & m_{2} & m \end{array} \right) = (-1)^{j+m+2j_{1}} \frac{1}{\sqrt{2j+1}}\langle j_{1} -m_{1} j_{2} -m_{2} | j m \rangle.
\end{equation}

Table of published $3j$ symbols [1] confirmed using this calculator.

| $j_{1}$ | $j_{2}$ | $j$ | $m_{1}$ | $m_{2}$ | $m$ | $3j$                 |
|---------|---------|-----|---------|---------|-----|----------------------|
| 1       | 1       | 1   | 1       | 0       |$-1$ | $-\frac{1}{\sqrt{6}}$|
| 3       | 3       | 3   | $-1$    | $-1$       |2 | 0|
|$\frac{7}{2}$ | $\frac{5}{2}$ | 2 | $\frac{7}{2}$ | $-\frac{3}{2}$ | $-2$ | $-\frac{\sqrt{2}}{6}$|
|$\frac{15}{2}$ | $\frac{15}{2}$ | 0 | $\frac{3}{2}$ | $-\frac{3}{2}$ | 0 | $\frac{1}{4}$|
|8 | $\frac{11}{2}$ | $\frac{9}{2}$ | 2 | $-\frac{7}{2}$ | $\frac{3}{2}$ | $-\frac{127}{12}\sqrt{\frac{6}{46189}}$|
| 8       | 7       | 6   | 3       | 0       |$-3$ | 0|

```python
print("Evaluated Wigner 3-j = {0}".format(am.symb3j(1,1,1,1,0,-1)))
print("Published Wigner 3-j = {0}".format(-1/np.sqrt(6)))
```

```python
print("Evaluated Wigner 3-j = {0}".format(am.symb3j(3,3,3,-1,-1,2)))
print("Published Wigner 3-j = {0}".format(0))
```

```python
print("Evaluated Wigner 3-j = {0}".format(am.symb3j(3.5,2.5,2,3.5,-1.5,-2)))
print("Published Wigner 3-j = {0}".format(-1*np.sqrt(2)/6))
```

```python
print("Evaluated Wigner 3-j = {0}".format(am.symb3j(7.5,7.5,0,1.5,-1.5,0)))
print("Published Wigner 3-j = {0}".format(1/4))
```

```python
print("Evaluated Wigner 3-j = {0}".format(am.symb3j(8,5.5,4.5,2,-3.5,1.5)))
print("Published Wigner 3-j = {0}".format((-127/12)*np.sqrt(6/46189)))
```

```python
print("Evaluated Wigner 3-j = {0}".format(am.symb3j(8,7,6,3,0,-3)))
print("Evaluated Wigner 3-j = {0} (5 d.p.)".format("%.5f"%am.symb3j(8,7,6,3,0,-3)))
print("Published Wigner 3-j = {0}".format(0))
```

## Calculations involving large integers

Although there are some limits, we have spot-checked a few results that are consistent with other evaluations based on different implementations involving calculations of angular momenta.  The evaluations of the following set of symbols involving large-integer angular momenta couplings have been confirmed using the `PyGammaRAD` calculators.

Clebsch-Gordan from Ref. [1]: $\langle 500~20~250~10~|~749~30 \rangle = 0$

```python
print("Evaluated CGC = {0}".format(am.cg(500,20,250,10,749,30)))
print("Published CGC = {0} [1]".format(0))
```

Wigner-3j from Ref. [1]: $\left( \begin{array}{ccc}
70 & 75 & 80\\
20 & -40 & 20 \end{array} \right) = -\frac{6347289054877299473594}{105} \times \sqrt{\frac{3116246}{159184087203947858242423601091756805001795494196699}} $

```python
eval_3j = am.symb3j(70,75,80,20,-40,20)
pub_3j = (-6347289054877299473594/105) * np.sqrt(3116246/159184087203947858242423601091756805001795494196699)

print("Evaluated Wigner 3-j = {0}".format(eval_3j))
print("Published Wigner 3-j = {0} [1]".format(pub_3j))
```

Wigner-3j from Ref. [4]: $\left( \begin{array}{ccc}
15 & 30 & 40\\
2 & 2 & -4 \end{array} \right) = -0.01908157979919155$

This result has also been confirmed using the calculator of Ref. [3], evaluated as: $-\frac{46874}{9} \times \sqrt{\frac{110}{8194888366823}}$

```python
eval_3j = am.symb3j(15,30,40,2,2,-4)
pub_3j_johannson = -0.01908157979919155
pub_3j_stone = -(46874/9)*np.sqrt(110/8194888366823)

print("Evaluated Wigner 3-j = {0}".format(eval_3j))
print("Published Wigner 3-j = {0} [4]".format(pub_3j_johannson))
print("Published Wigner 3-j = {0} [3]".format(pub_3j_stone))
```

Wigner-6j from Ref. [1]: $\left\{ \begin{array}{ccc}
80 & 70 & 60\\
60 & 70 & 80 \end{array} \right\} = \frac{22937084180342998024692848841405644906343471096259}{3807942910488176294742031868978400502934196590236245} \times \sqrt{\frac{133733857259383840245}{65742467686458947573719}} $

# Tables of angular distribution coefficients from Rose and Brink

Use this `Notebook` to display the original tabulated data from Rose and Brink [1] and compare with the corresponding coefficients calculated using the `PyGammaRAD` library.

```python
import PyGammaRAD as pg
am = pg.AngularMomentum()
import numpy as np
import pandas as pd
import logging
```

## $R_{k}(L_{1}L_{2}J_{i}J_{f})$ coefficients

These tables are given on p. 338-340 in the Appendix of Ref. [1].  The $R_{k}$ angular-distribution coefficient is defined in Eqs. (3.36) and (3.37) of Ref. [1].  Note that $L_{2} = L_{1} + 1$.

```python
df_R_int = am.get_tableRa()
df_R_int
```

```python
df_R_halfint = am.get_tableRb()
df_R_halfint
```

## $U_{k}(L_{1}J_{i}J_{f})$ and $U_{k}(L_{2}J_{i}J_{f})$ coefficients

These tables are given on p. 341-342 in the Appendix of Ref. [1].  The $U_{k}$ angular-distribution coefficient is defined in Eq. (3.45) of Ref. [1]; this expression is equivalent to the expression given by Yamazaki in Eq. (14) from Ref. [2].  Note that $L_{2} = L_{1} + 1$.

```python
df_U = am.get_tableU()
df_U
```

## $S_{k}(l_{1} l_{2}J s)$ coefficients

These tables are given on p. 343-345 in the Appendix of Ref. [1].  The $S_{k}$ population-tensor coefficient is defined in Eq. (3.59) of Ref. [1].

```python
df_S_int = am.get_tableSa()
df_S_int
```

```python
df_S_halfint = am.get_tableSb()
df_S_halfint
```

## $\rho_{k}(J m)$ coefficients

These tables are given on p. 346-347 in the Appendix of Ref. [1].  The $\rho_{k}$ statistical-tensor coefficient is defined in Eq. (3.63) of Ref. [1].

```python
df_p_int = am.get_tablePa()
df_p_int
```

```python
df_p_halfint = am.get_tablePb()
df_p_halfint
```

## Exporting Rose and Brink's data tables to file

All tables from the Appendix of Ref. [2] can all be exported to a convenient CSV or JSON format in the current working directory by calling the `rose_brink2file` method and passing the appropriate format argument as explained in the docstring.

```python
help(am.rose_brink2file)
```

```python
# Print Table R (integral J) to CSV: dumps RoseBrinkTableRa.csv in current working directory
am.rose_brink2file('RA','CSV')
# Print Table R (integral J) to JSON: dumps RoseBrinkTableRa.json in current working directory
am.rose_brink2file('RA','JSON')
```

```python
# Print Table R (half-integral J) to CSV: dumps RoseBrinkTableRb.csv in current working directory
am.rose_brink2file('RB','CSV')
# Print Table R (half-integral J) to JSON: dumps RoseBrinkTableRb.json in current working directory
am.rose_brink2file('RB','JSON')
```

```python
# Print Table U to CSV: dumps RoseBrinkTableU.csv in current working directory
am.rose_brink2file('U','CSV')
# Print Table U to JSON: dumps RoseBrinkTableU.json in current working directory
am.rose_brink2file('U','JSON')
```

```python
# Print Table S (integral J) to CSV: dumps RoseBrinkTableSa.csv in current working directory
am.rose_brink2file('SA','CSV')
# Print Table S (integral J) to JSON: dumps RoseBrinkTableSa.json in current working directory
am.rose_brink2file('SA','JSON')
```

```python
# Print Table R (half-integral J) to CSV: dumps RoseBrinkTableSb.csv in current working directory
am.rose_brink2file('SB','CSV')
# Print Table R (half-integral J) to JSON: dumps RoseBrinkTableSb.json in current working directory
am.rose_brink2file('SB','JSON')
```

```python
# Print Table p (integral J) to CSV: dumps RoseBrinkTablePa.csv in current working directory
am.rose_brink2file('PA','CSV')
# Print Table p (integral J) to JSON: dumps RoseBrinkTablePa.json in current working directory
am.rose_brink2file('PA','JSON')
```

```python
# Print Table p (half-integral J) to CSV: dumps RoseBrinkTablePb.csv in current working directory
am.rose_brink2file('PB','CSV')
# Print Table p (half-integral J) to JSON: dumps RoseBrinkTablePb.json in current working directory
am.rose_brink2file('PB','JSON')
```

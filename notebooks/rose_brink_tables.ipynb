# Tables of angular distribution coefficients from Rose and Brink

Use this `Notebook` to display the original tabulated data from Rose and Brink [1] and compare with the corresponding coefficients calculated using the `PyGammaRAD` library.

```python
import PyGammaRAD as pg
am = pg.AngularMomentum()
import numpy as np
import pandas as pd
import logging
```

## $R_{k}(L_{1}L_{2}J_{i}J_{f})$ coefficient

These tables are given on p. 338-340 in the Appendix of Ref. [1].  The $R_{k}$ coefficient is defined in Eqs. (3.36) and (3.37) of Ref. [1].  Note that $L_{2} = L_{1} + 1$.

```python
df_R_int = am.get_tableRa()
df_R_int
```

```python
df_R_halfint = am.get_tableRb()
df_R_halfint
```

## $U_{k}(L_{1}J_{i}J_{f})$ and $U_{k}(L_{2}J_{i}J_{f})$ coefficients

These tables are given on p. 341-342 in the Appendix of Ref. [1].  The $U_{k}$ coefficient is defined in Eq. (3.45) of Ref. [1]; this expression is equivalent to the expression given by Yamazaki in Eq. (14) from Ref. [2].  Note that $L_{2} = L_{1} + 1$.

```python
df_U = am.get_tableU()
df_U
```

[2] T. Yamazaki, *Tables of Coefficients for Angular Distribution of Gamma Rays from Aligned Nuclei*, Nucl. Data Sect. A **3**, 1 (1967).

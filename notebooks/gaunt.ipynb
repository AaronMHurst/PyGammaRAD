# Calculation of Gaunt coefficients

Use this `Notebook` to compare the calculated Gaunt coefficients published by Silay Yukcu [1] with the corresponding results obtained using the `gaunt` method of the `PyGammaRAD` library.

```python
import PyGammaRAD as pg
am = pg.AngularMomentum()
import numpy as np
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.max_row', None)
```

## Selection rules for Gaunt coefficients

Check quantum numbers of Table 1 in Ref. [1] satisfy the selection rules, namely:

(i) **SR1** Triangle condition: $| l_{1} - l_{2} | \leq l_{3} \leq l_{1} + l_{2}$.

(ii) **SR2** $(l_{1} + l_{2} + l_{3}) \mod 2 = 0$.

(iii) **SR3** $m_{1} + m_{2} + m_{3} = 0$.


```python
# Read original Gaunt-coefficient test data from Yukcu into DataFrame
gaunt_df = pd.read_csv("test_csv_data/yukcu_gaunt_original.csv")
#gaunt_df
```

```python
# Create DataFrame containing output of selection rules based on input quantum numbers explicitly given in Table 1 [1]
gaunt_list = gaunt_df.values.tolist()
gaunt_column_names = ['l1','m1','l2','m2','l3','m3','SR1','SR2','SR3']
gaunt_rows = []
for QN in gaunt_list:
    l1 = QN[0]
    m1 = QN[1]
    l2 = QN[2]
    m2 = QN[3]
    l3 = QN[4]
    m3 = QN[5]

    # SR1 |l1-l2| <= l3 <= l1+l2
    SR1 = False
    if (abs(l1-l2) <= l3) and (l3 <= (l1+l2)):
        SR1 = True
    else:
        SR1 = False
    #print("SR1",abs(l1-l2), l3, l1+l2, SR1)

    # SR2 (l1+l2+l3) mod 2 = 0 
    SR2 = False
    L = l1+l2+l3
    if L%2==0:
        SR2 = True
    else:
        SR2 = False
    #print("SR2", L%2, SR2)

    # SR3 m1+m2+m3 = 0
    SR3 = False
    M = m1+m2+m3
    if M==0:
        SR3 = True
    else:
        SR3 = False
    #print("SR3", M, SR3)

    gaunt_rows.append([l1,m1,l2,m2,l3,m3, SR1, SR2, SR3])
selection_rules_df = pd.DataFrame(gaunt_rows, columns=gaunt_column_names)
selection_rules_df
```

Based on the original data of Table 1 in Ref. [1] the above results show that:

(i) **SR1** is satisfied for all cases.

(ii) **SR2** is satisfied for all cases.

(iii) **SR3** is not satisfied many cases.

Based on these observations, we have adjusted the signs on some of the $m$ quantum numbers in Table 1 of Ref. [1] to ensure that **SR3** is satisfied in all cases.  In the adjusted dataset we have also removed the row of data corresponding to index #`5` in the above `DataFrame` which appears to be a duplicate.

```python
# Read original Gaunt-coefficient test data from Yukcu into DataFrame
gaunt_df = pd.read_csv("test_csv_data/yukcu_gaunt_adjusted.csv")
#gaunt_df
```

```python
# Create DataFrame containing output of selection rules based on adjusted input quantum numbers given in Table 1 [1]
gaunt_list = gaunt_df.values.tolist()
gaunt_column_names = ['l1','m1','l2','m2','l3','m3','SR1','SR2','SR3']
gaunt_rows = []
for QN in gaunt_list:
    l1 = QN[0]
    m1 = QN[1]
    l2 = QN[2]
    m2 = QN[3]
    l3 = QN[4]
    m3 = QN[5]

    # SR1 |l1-l2| <= l3 <= l1+l2
    SR1 = False
    if (abs(l1-l2) <= l3) and (l3 <= (l1+l2)): SR1 = True
        
    # SR2 (l1+l2+l3) mod 2 = 0 
    SR2 = False
    L = l1+l2+l3
    if L%2==0: SR2 = True
    
    # SR3 m1+m2+m3 = 0
    SR3 = False
    M = m1+m2+m3
    if M==0: SR3 = True
    
    gaunt_rows.append([l1,m1,l2,m2,l3,m3, SR1, SR2, SR3])
selection_rules_df = pd.DataFrame(gaunt_rows, columns=gaunt_column_names)
selection_rules_df
```

```python
# We can now evaluate the Gaunt coefficients using the adjusted dataset and compare to results of Table [1]:
# (i) Most results agree numerically to within a reasonable level of accuracy.
# (ii) There are a few differences in signs; however our signs agree with some other methods such as SymPy used for Gaunt calculations.
gaunt_column_names = ['l1','m1','l2','m2','l3','m3','G']
gaunt_rows_results = []
for QN in gaunt_list:
    l1 = QN[0]
    m1 = QN[1]
    l2 = QN[2]
    m2 = QN[3]
    l3 = QN[4]
    m3 = QN[5]

    G = am.gaunt(l1,l2,l3,m1,m2,m3)
    gaunt_rows_results.append([l1,m1,l2,m2,l3,m3,'%.14e'%G])
gaunt_results_df = pd.DataFrame(gaunt_rows_results, columns=gaunt_column_names)
gaunt_results_df
```

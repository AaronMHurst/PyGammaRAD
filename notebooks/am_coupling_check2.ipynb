# Angular momentum coupling coefficients

Use this `Notebook` to compare Clebsch-Gordan coefficients, Racah coefficients, and Wigner $9$-$j$ symbols calculated by Taro Tamura and published in the *TEST RUN OUTPUT* of the associated publication [1] with the corresponding results obtained using `PyGammaRAD` methods.  Wigner $9$-$j$ results calculated and published by Liqiang Wei [2] are also compared in this `Notebook` along with symbolic Wigner $3$-$j$ symbols published by Brink and Satchler [3].

```python
import PyGammaRAD as pg
am = pg.AngularMomentum()
import numpy as np
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.max_row', None)
```

##  TEST RUN OUTPUT: Clebsch-Gordan Coefficients

Check `DataFrame` results against documented output on p. 341 [1].  To calculate the Clebsch-Gordan coefficient using Eq. (1) of Ref. [1], the authors adopt the following convention:

`IA` = 2$j_{1}$, `IB` = 2$j_{2}$, `IC` = 2$J$,

`ID` = 2$m_{1}$, `IE` = 2$m_{2}$, `IF` = 2$M$.

Accordingly, the `PyGammaRAD` method `cg` then takes the input as `cg(j1, m1, j2, m2, J, M)` in order to calculate the corresponding Clebsch-Gordan coefficient.

```python
# Read Clebsch-Gordan test data from Tamura into DataFrame
cleb_df = pd.read_csv("test_csv_data/tamura_cleb_test.csv")
cleb_list = cleb_df.values.tolist()
```

```python
# Create new DataFrame containing evaluated Clebsch-Gordan coefficients based on Tamura's input data
cg_column_names = ['IA', 'IB', 'IC', 'ID', 'IE', 'IF', 'C.G. COEFF']
cg_results_rows = []
for IX in cleb_list:
    IA = IX[0]
    IB = IX[1]
    IC = IX[2]
    ID = IX[3]
    IE = IX[4]
    IF = IX[5]

    j1 = IA/2
    j2 = IB/2
    J = IC/2
    m1 = ID/2
    m2 = IE/2
    M = IF/2

    cg_coeff = am.cg(j1,m1,j2,m2,J,M)
    #print(IA,IB,IC,ID,IE,IF,"%.6e"%cg_coeff)
    cg_results_rows.append([IA,IB,IC,ID,IE,IF,"%.6e"%cg_coeff])
cg_results_df = pd.DataFrame(cg_results_rows, columns=cg_column_names)
print(cg_results_df)
```

##  TEST RUN OUTPUT: Racah Coefficients

Check `DataFrame` results against documented output on p. 341-342 [1].  To calculate the Racah coefficient using Eq. (2) of Ref. [1], the authors adopt the following convention:

`IA` = 2$a$, `IB` = 2$b$, `IC` = 2$c$,

`ID` = 2$d$, `IE` = 2$e$, `IF` = 2$f$.

Accordingly, the `PyGammaRAD` method `racah` then takes the input as `racah(a, b, c, d, e, f)` in order to calculate the corresponding Racah coefficient.

```python
# Read Racah test data from Tamura into DataFrame
racah_df = pd.read_csv("test_csv_data/tamura_rac_test.csv")
racah_list = racah_df.values.tolist()
```

```python
# Create new DataFrame containing evaluated Racah coefficients based on Tamura's input data
rac_column_names = ['IA', 'IB', 'IC', 'ID', 'IE', 'IF', 'RACAH COEFF']
rac_results_rows = []
for i,IX in enumerate(racah_list):
    
    IA = IX[0]
    IB = IX[1]
    IC = IX[2]
    ID = IX[3]
    IE = IX[4]
    IF = IX[5]

    a = IA/2
    b = IB/2
    c = IC/2
    d = ID/2
    e = IE/2
    f = IF/2

    rac_coeff = am.racah(a,b,c,d,e,f)
    #print(IA,IB,IC,ID,IE,IF,"%.6e"%rac_coeff)
    rac_results_rows.append([IA,IB,IC,ID,IE,IF,"%.6e"%rac_coeff])
rac_results_df = pd.DataFrame(rac_results_rows, columns=rac_column_names)
print(rac_results_df)
```

##  TEST RUN OUTPUT: Wigner 9-$j$ symbols

Check `DataFrame` results against documented output on p. 342 [1].  To evaluate the Wigner 9-$j$ symbol using Eq. (3) of Ref. [1], the authors adopt the following convention: `L9(N)` = 2$j_{n}$, thus,

`L1` = 2$j_{1}$, `L2` = 2$j_{2}$, `L3` = 2$j_{4}$, `L4` = 2$j_{5}$, `L5` = 2$j_{3}$, `L6` = 2$j_{6}$, `L7` = 2$j_{7}$, `L8` = 2$j_{8}$, `L9` = 2$j_{9}$.


Note the `L`/$j$ ordering with respect to the symbol defined in Eq. (3) [1] for input arguments $j_{5}$, $j_{3}$, and $j_{4}$.  Accordingly, the `PyGammaRAD` method `symb9j` then takes the input as `symb9j(j1,j2,j3,j4,j5,j6,j7,j8,j9)` in order to evaluate the corresponding Wigner 9-$j$ symbol.

```python
# Read Wigner 9-j test data from Tamura into DataFrame
wigner9j_df = pd.read_csv("test_csv_data/tamura_9j_test.csv")
wigner9j_list = wigner9j_df.values.tolist()
```

```python
# Create new DataFrame containing evaluated Wigner 9-j symbols based on Tamura's input data
wigner9j_column_names = ['L1', 'L2', 'L3', 'L4', 'L5', 'L6', 'L7', 'L8', 'L9', 'WIGNER-9j SYMB']
wigner9j_results_rows = []
for i,IX in enumerate(wigner9j_list):
    
    L1 = IX[0]
    L2 = IX[1]
    L3 = IX[2]
    L4 = IX[3]
    L5 = IX[4]
    L6 = IX[5]
    L7 = IX[6]
    L8 = IX[7]
    L9 = IX[8]

    j1 = L1/2
    j2 = L2/2
    j3 = L5/2
    j4 = L3/2
    j5 = L4/2
    j6 = L6/2
    j7 = L7/2
    j8 = L8/2
    j9 = L9/2

    wigner9j_symb = am.symb9j(j1,j2,j3,j4,j5,j6,j7,j8,j9)
    #print(L1,L2,L3,L4,L5,L6,L7,L8,L9,"%.6e"%wigner9j_symb)
    wigner9j_results_rows.append([L1,L2,L3,L4,L5,L6,L7,L8,L9,"%.6e"%wigner9j_symb])
wigner9j_results_df = pd.DataFrame(wigner9j_results_rows, columns=wigner9j_column_names)
print(wigner9j_results_df)
```

## Further Wigner 9-$j$ tests

Use the following cells to test Wigner 9-$j$ results tabulated in Ref. [2] and compare execution times.

```python
import time
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', 120)
```

```python
# Read Wigner 9-j test data from Lei into DataFrame
wigner9j_lei_df = pd.read_csv("test_csv_data/wei_9j_test.csv")
wigner9j_lei_list = wigner9j_lei_df.values.tolist()
```

```python
wigner9j_lei_column_names = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'j', 'WIGNER-9j SYMB', 'CPU TIME [s]']
wigner9j_lei_results_rows = []
for i,IX in enumerate(wigner9j_lei_list):
    
    a = IX[0]
    b = IX[1]
    c = IX[2]
    d = IX[3]
    e = IX[4]
    f = IX[5]
    g = IX[6]
    h = IX[7]
    j = IX[8]

    start_time = time.perf_counter()
    wigner9j_symb = am.symb9j(a,b,c,d,e,f,g,h,j)
    stop_time = time.perf_counter()
    elapsed_time = stop_time - start_time
    #print(a,b,c,d,e,f,g,h,j,"%.15e"%wigner9j_symb,"%.7e"%elapsed_time)
    wigner9j_lei_results_rows.append([a,b,c,d,e,f,g,h,j,"%.15e"%wigner9j_symb,"%.7e"%elapsed_time])
wigner9j_lei_results_df = pd.DataFrame(wigner9j_lei_results_rows, columns=wigner9j_lei_column_names)
print(wigner9j_lei_results_df)
```

# Yamazaki's tables of angular distribution coefficients

Use this `Notebook` to display the original tabulated data from Yamazaki [1] and compare with the corresponding coefficients calculated using methods from the `PyGammaRAD` package.

```python
import PyGammaRAD as pg
am = pg.AngularMomentum()
import numpy as np
import pandas as pd
import logging
```

```python
def compare_table1(df):
    """Method to generate list of comparative values of statistical
    population tensors from Table 1 in Yamazaki's paper [1].
    
    Arguments:
        df: DataFrame object returned from `get_table1` method.
    
    Returns:
        A list object comparing the corresponding tabulated and
        calculated statistical population tensors."""
    
    B_data = []
    ldf_B = df.values.tolist()
    for row in ldf_B:
        J, m = row[0], row[1]
        B2_table, B4_table, B6_table = row[2], row[3], row[4]
        B2_calc, B4_calc, B6_calc = 0.0, 0.0, 0.0
    
        B2_calc = float("%.5f"%am.calc_B(2,J))
        if abs(B4_table) > 0.0:
            B4_calc = float("%.5f"%am.calc_B(4,J))
        if abs(B6_table) > 0.0:
            B6_calc = float("%.5f"%am.calc_B(6,J))
    
        B_data.append([J, m, B2_table, B2_calc, B4_table, B4_calc, B6_table, B6_calc])
    return B_data

def compare_table2(df):
    """Method to generate list of comparative values of coefficients
    from Table 2(a) and (b) in Yamazaki's paper [1].
    
    Arguments:
        df: DataFrame object returned from `get_table2a`
            or `get_table2b` methods.
    
    Returns:
        A list object comparing the corresponding tabulated and
        calculated coefficients."""
    
    angdist_data = []
    ldf_angdist = df.values.tolist()
    for row in ldf_angdist:
        Ji = row[0]
        Jf = row[1]
        L1 = row[2]
        L2 = row[3]
        
        F2_table = row[4]
        B2F2_table = row[5]
        F4_table = row[6]
        B4F4_table = row[7]    
        U2_table = row[8]
        U4_table = row[9]

        #F2_calc, B2F2_calc = 0.0, 0.0
        F4_calc, B4F4_calc = float(0.0), float(0.0)
        U2_calc, U4_calc = float(0.0), float(0.0)

        F2_calc = float("%.5f"%am.calc_F(2,Jf,L1,L2,Ji))
        B2F2_calc = float("%.5f"%am.calc_BF(2,Jf,L1,L2,Ji))

        if abs(F4_table) > 0:
            F4_calc = float("%.5f"%am.calc_F(4,Jf,L1,L2,Ji))
        if abs(B4F4_table) > 0:
            B4F4_calc = float("%.5f"%am.calc_BF(4,Jf,L1,L2,Ji))

        if (abs(U2_table) > 0) and (int(L1)==int(L2)):
            U2_calc = float("%.5f"%am.calc_u(2, Ji, L1, Jf))
        if (abs(U4_table) > 0) and (int(L1)==int(L2)):
            U4_calc = float("%.5f"%am.calc_u(4, Ji, L1, Jf))

        if ((2*Ji)%2==0) and ((2*Jf)%2==0):
            Ji = int(Ji)
            Jf = int(Jf)
        
        angdist_data.append([Ji,Jf,int(L1),int(L2), 
                             F2_table,F2_calc, B2F2_table,B2F2_calc,
                             F4_table,F4_calc, B4F4_table,B4F4_calc,
                             U2_table,U2_calc,U4_table,U4_calc])
    return angdist_data
```

## Maximum alignment anisotropy coefficients: $A_{k}^{max}$

The anisotropy coefficent $A_{k}$ assuming complete nuclear alignment can be calculated using the `A_max` method.  For $J_{i} \rightarrow J_{f}$ transitions of pure multipole order $L_{1} = L_{2} = L$ there is no mixing ratio, i.e., $\delta_{\gamma} = 0$, and the $A_{k}$ coefficient reduces to:

$$ A_{k}^{\text{max}}(J_{i}L_{1}L_{2}J_{f}) \equiv A_{k}^{\text{max}}(J_{i}L L J_{f}) = f_{k}(J_{f}L L J_{i}),$$ 

where

$$f_{k}(J_{f}L L J_{i}) \equiv B_{k}(J_{i}) F_{k}(J_{f}L L J_{i}).$$

It is therefore useful to compare maximum alignment $A_{k}$ results from these pure-multipole cases to tabulated data in Table 2 [1], and thus test to ensure the relationship $A_{k}^{\text{max}}(J_{i}L L J_{f}) = B_{k}(J_{i}) F_{k}(J_{f}L L J_{i})$ holds.

```python
help(am.A_max)
```

```python
def compare_A_max(df):
    """Method to compare calculated anisotropy coefficients based on 
    complete nuclear alignment to BkFk coefficients listed in Tables 
    2(a) and 2(b) from Yamazaki's paper [1].  
    
    Arguments:
        df: DataFrame object based on DataFrames generated in Sections 
            'Table 2(a)' and 'Table 2(b)' of this Notebook.  Pass either:
            `df_angint_comp` integral J;
            `df_anghalf_comp` half-integral J.
            
    Returns:
        A list object comparing the corresponding tabulated and
        calculated coefficients used to determine the calculated  
        anisotropy coefficients A2 and A4."""
    ldf = df.values.tolist()
    A_max_data = []
    for row in ldf:
        Ji = row[0]
        Jf = row[1]
        L1 = row[2]
        L2 = row[3]
    
        F2_table = row[4]
        B2F2_table = row[6]
        F4_table = row[8]
        B4F4_table = row[10]
    
        if int(L1) == int(L2):
        
            B2_calc, F2_calc, B2F2_calc, A2_max = 0.0, 0.0, 0.0, 0.0
            B4_calc, F4_calc, B4F4_calc, A4_max = 0.0, 0.0, 0.0, 0.0
            if abs(float(F2_table))>0.0 and abs(float(B2F2_table))>0.0:
                # k = 2
                B2_calc = am.calc_B(2,Ji)
                F2_calc = am.calc_F(2,Jf,L1,L2,Ji)
                B2F2_calc = B2_calc*F2_calc        
                A2_max = am.A_max(2,Ji,L1,L2,Jf)
        
                try:
                    B4_calc = am.calc_B(4,Ji)
                #except ValueError:
                except TypeError:
                    B4_calc = 0.0
                
                if abs(float(F4_table))>0.0 and abs(float(B4F4_table))>0.0:
                    # k = 4
                    #B4_calc = am.calc_B(4,Ji)
                    F4_calc = am.calc_F(4,Jf,L1,L2,Ji)
                    B4F4_calc = B4_calc*F4_calc        
                    A4_max = am.A_max(4,Ji,L1,L2,Jf)
            
                if (2*Ji)%2==0 and (2*Jf)%2==0:
                    Ji = int(Ji)
                    Jf = int(Jf)    
            
                A_max_data.append([Ji,Jf,int(L1),int(L2),
                               F2_table,B2F2_table, B2_calc, B2F2_calc, A2_max,
                               F4_table,B4F4_table, B4_calc, B4F4_calc, A4_max])
                
    return A_max_data
```

```python
anisotropy_columns = ['Ji', 'Jf', 'L1', 'L2', 
                     'F2 (table)', 'B2F2 (table)', 'B2 (calc)', 'B2*F2 (calc)', 'A2 max (dg=0)',
                     'F4 (table)', 'B4F4 (table)', 'B4 (calc)', 'B4*F4 (calc)', 'A4 max (dg=0)']
```

## Angular distribution coefficients: $U_{k}$

The coefficent $U_{k}$ can be calculated using the `U_coeff` method.  For $J_{i} \rightarrow J_{f}$ transitions of pure multipole order $L_{1} = L_{2} = L$ there is no mixing ratio, i.e., $\delta_{\gamma} = 0$, and the $U_{k}$ coefficient reduces to:

$$ U_{k}(J_{i}L_{1}L_{2}J_{f}) \equiv U_{k}(J_{i}L L J_{f}) = u_{k}(J_{i}L J_{f}).$$ 

It is therefore useful to compare $U_{k}$ coefficients calculated for these pure-multipole cases to tabulated data in Table 2 [1] for $u_{k}$, and thus test to ensure the above relationship $U_{k}(J_{i}L L J_{f}) = u_{k}(J_{i} L J_{f})$ holds.

```python
help(am.U_coeff)
```

```python
def compare_U(df):
    """Method to compare calculated angular distribution coefficients Uk 
    for transitions of pure multipolarity to uk coefficients listed in 
    Tables 2(a) and 2(b) from Yamazaki's paper [1].  
    
    Arguments:
        df: DataFrame object based on DataFrames generated in Sections 
            'Table 2(a)' and 'Table 2(b)' of this Notebook.  Pass either:
            `df_angint_comp` integral J;
            `df_anghalf_comp` half-integral J.
            
    Returns:
        A list object comparing the corresponding tabulated and
        calculated angular distribution coefficients for uk and Uk."""
    ldf = df.values.tolist()
    U_data = []
    for row in ldf:
        Ji = row[0]
        Jf = row[1]
        L1 = row[2]
        L2 = row[3]
    
        U2_table = row[12]
        U4_table = row[14]
    
        if int(L1) == int(L2):
        
            U2 = 0.0
            U4 = 0.0
            if abs(float(U2_table))>0.0:
                # k = 2 
                U2 = am.U_coeff(2,Ji,L1,L2,Jf)
                
                if abs(float(U4_table))>0.0:
                    # k = 4
                    #print(U4_table)
                    U4 = am.U_coeff(4,Ji,L1,L2,Jf)
            
                if (2*Ji)%2==0 and (2*Jf)%2==0:
                    Ji = int(Ji)
                    Jf = int(Jf)    
            
                U_data.append([Ji, Jf, int(L1), int(L2), U2_table, U2, U4_table, U4])
                
    return U_data
```

```python
u_columns = ['Ji', 'Jf', 'L1', 'L2', 'u2 (table)', 'U2', 'u4 (table)', 'U4']
```

```python
# Integral spin U angular distributions
with pg.LogLevelContext(logging.WARNING):
    U_integral_data = compare_U(df_angint_comp)
```

```python
df_U_integral = pd.DataFrame(U_integral_data, columns=u_columns)
df_U_integral
```

```python
# Half-integral spin U angular distributions
with pg.LogLevelContext(logging.WARNING):
    U_halfint_data = compare_U(df_anghalf_comp)
```

```python
df_U_halfint = pd.DataFrame(U_halfint_data, columns=u_columns)
df_U_halfint
```

# Angular distributions

Use this `Notebook` to illustrate the angular distribution based on anisotropy coefficients ($A_{k}$) calculated using the `PyGammaRAD` package.  The angular distribution for a transition from $J_{i} \rightarrow J_{f}$ may be expressed as

$$W(\theta) = 1 + A_{2}P_{2}(\cos \theta) + A_{4}P_{4}(\cos \theta).$$

The Legendre polynomials $P_{k}$ can be derived from Rodrigues' formula which, given in terms of $\cos \theta$, may be written as:

\begin{equation}
P_{k}(\cos \theta) = \frac{1}{2^{k}k!} \frac{d^{k}}{d\cos^{k}\theta} (\cos^{2} \theta - 1)^{k}.
\end{equation}

Accordingly, the first three even-$k$ terms that are needed to describe $W(\theta)$ may then be expressed as:

$$P_{0}(\cos \theta) = 1$$

$$P_{2}(\cos \theta) = \frac{1}{2} (3 \cos^{2}\theta -1)$$

$$P_{4}(\cos \theta) = \frac{1}{8} (35 \cos^{4}\theta - 30 \cos^{2} +3).$$

````python
import PyGammaRAD as pg
am = pg.AngularMomentum()
import numpy as np
import matplotlib.pyplot as plt
%matplotlib notebook
````

````python
def deg2rad(x):
    return (x/180.0) * np.pi

def lpoly0(x):
    #return x/x
    np.seterr(divide='ignore', invalid='ignore')
    return np.where(x!=0, np.divide(x,x), 1.0)
    
def lpoly2(x):
    return 0.5 * (3*np.cos(deg2rad(x))**2 - 1)

def lpoly4(x):
    return (1.0/8.0) * ((35*np.cos(deg2rad(x))**4) - (30*np.cos(deg2rad(x))**2) + 3)

def W_theta_dist(A2,A4,x):
    P2 = lpoly2(x)
    P4 = lpoly4(x)
    W = 1 + (A2*P2) + (A4*P4)
    return W
````

````python
%matplotlib notebook

theta = np.linspace(0.0, 180.0, 1800)
p0 = lpoly0(theta)
p2 = lpoly2(theta)
p4 = lpoly4(theta)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(9,5))
ax1.plot(np.cos(deg2rad(theta)),p0,color='gray',lw=2,linestyle=':',label=r'$P_{0}(\cos\theta)$')
ax1.plot(np.cos(deg2rad(theta)),p2,color='green',lw=2,linestyle='--',label=r'$P_{2}(\cos\theta)$')
ax1.set_xlabel(r'$\cos(\theta)$', size=20)
ax1.set_ylabel(r'$P_{k}(\cos(\theta))$', size=20)
ax1.tick_params(axis='x', labelsize=16)
ax1.tick_params(axis='y', labelsize=16)
ax1.legend(loc='lower right',shadow=True,framealpha=0.1,borderpad=0.5,prop={'size':14})

ax2.plot(np.cos(deg2rad(theta)),p4,color='orange',lw=2,linestyle='--',label=r'$P_{4}(\cos\theta)$')
ax2.set_xlabel(r'$\cos(\theta)$', size=20)
ax2.set_ylabel(r'$P_{4}(\cos(\theta))$', size=20)
ax2.tick_params(axis='x', labelsize=16)
ax2.tick_params(axis='y', labelsize=16)
ax2.legend(loc='lower right',shadow=True,framealpha=0.1,borderpad=0.5,prop={'size':14})

plt.tight_layout()
plt.show()
````

````python
# Calculate the anisotropy coefficients A2 and A4 assuming maximum (complete) nuclear alignment
# Example: Ji=2 -> Jf=0 transition where L1=L2=2 for a pure stretched E2 quadrupole (mixing ratio dg=0)
A2 = am.A_max(2,2,2,2,0)
A4 = am.A_max(4,2,2,2,0)
W_20 = W_theta_dist(A2,A4,theta)
print("Ji=2+ -> Jf=0+")
print("A2 = {0}".format(A2))
print("A4 = {0}".format(A4))
````

````python
# Calculate the anisotropy coefficients A2 and A4 assuming maximum (complete) nuclear alignment
# Example: Ji=4 -> Jf=2 transition where L1=L2=2 for a pure stretched E2 quadrupole (mixing ratio dg=0)
A2 = am.A_max(2,4,2,2,2)
A4 = am.A_max(4,4,2,2,2)
W_42 = W_theta_dist(A2,A4,theta)
print("Ji=4+ -> Jf=2+")
print("A2 = {0}".format(A2))
print("A4 = {0}".format(A4))
````

````python
# Calculate the anisotropy coefficients A2 and A4 assuming maximum (complete) nuclear alignment
# Example: Ji=2 -> Jf=2 transition where L1=1 and L2=2 for a mixed M1+E2 (mixing ratio dg=-0.18)
A2, A4 = 0.0, 0.0
#try:
A2 = am.A_max(2,2,1,2,2,-0.18)
A4 = am.A_max(4,2,1,2,2,-0.18)
#except ValueError:
#except TypeError:
#except AssertionError:
#    if not A2:
#        A2 = 0.0
#    if not A4:
#        A4 = 0.0
W_22 = W_theta_dist(A2,A4,theta)
print("Ji=2+ -> Jf=2+")
print("A2 = {0}".format(A2))
print("A4 = {0}".format(A4))
````

## Comparison with experimental data

The theoretical anisotropy coefficients, $A_{2}$ and $A_{4}$, described above have been calculated assuming maximum nuclear alignment.  This is not always a reliable approximation in real experimental scenarios where the alignment is only partial and an anisotropy-attenuation coefficient $\alpha_{k}$ is needed, where

$$ \alpha_{k} = \frac{\rho_{k}(J)}{B_{k}(J)}.$$

Here, $\rho_{k}(J)$ is the statistical popoulation tensor according to the formation of the state based on experimental conditions, and $B_{k}(J)$ is the corresponding statistical population tensor for the state assuming complete nuclear alignment.  The experimental anisotropy coefficient corrected for alignment attenuation may then be expressed as

$$ a_{k} = \alpha_{k} A_{k}.$$

````python
import pandas as pd
from scipy.interpolate import UnivariateSpline as US
````

````python
# Load experimental a2 and a4 data for 847-keV 2+->0+ in 56Fe
df_a2_847 = pd.read_csv("expt_csv_data/a2_gamma_847.csv")
print(df_a2_847.head())
e2_847 = df_a2_847['E'].values.tolist()
a2_847 = df_a2_847['a2'].values.tolist()

df_a4_847 = pd.read_csv("expt_csv_data/a4_gamma_847.csv")
print(df_a4_847.head())
e4_847 = df_a4_847['E'].values.tolist()
a4_847 = df_a4_847['a4'].values.tolist()
````

````python
# Define linear space for interpolation
E_interp = np.linspace(0.8, 15.00, 1000)
````

````python
# Plot a2 and a4 coefficients for 847-keV gamma and fit cubic-spline to data
%matplotlib notebook

fig, (ax1, ax2) = plt.subplots(1,2, figsize=(9,5))
ax1.scatter(e2_847,a2_847, label=r'$a_{2}: 847$ keV', color='magenta')
ax1.set_xlabel(r'$E_{n}$ [MeV]', size=16)
ax1.set_ylabel(r'$a_{2}$', size=16)
ax1.tick_params(axis='x', labelsize=16)
ax1.tick_params(axis='y', labelsize=16)
#ax1.set_ylim(top=0.62)
ax1.grid(True)


ax2.scatter(e4_847,a4_847, label=r'$a_{4}: 847$ keV', color='cyan')
ax2.set_xlabel(r'$E_{n}$ [MeV]', size=16)
ax2.set_ylabel(r'$a_{4}$', size=16)
ax2.tick_params(axis='x', labelsize=12)
ax2.tick_params(axis='y', labelsize=12)
ax2.grid(True)

order = 3
spl2_847 = US(e2_847,a2_847, k=order)
spl4_847 = US(e4_847,a4_847, k=order)

spl2_847.set_smoothing_factor(0.02)
spl4_847.set_smoothing_factor(0.02)
ax1.plot(E_interp, spl2_847(E_interp), 'k', lw=3, label='cubic spline')
ax2.plot(E_interp, spl4_847(E_interp), 'k', lw=3, label='cubic spline')

ax1.legend(loc='best',fancybox=True,framealpha=0.5,shadow=False,borderpad=0.5,prop={'size':12})
ax2.legend(loc='best',fancybox=True,framealpha=0.5,shadow=False,borderpad=0.5,prop={'size':12})

plt.tight_layout()
plt.show()
````
